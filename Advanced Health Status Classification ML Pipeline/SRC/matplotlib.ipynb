# Bar Graph - Mean,Sum,Range

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

e = pd.read_csv("/kaggle/input/video-game-sales-with-ratings/Video_Games_Sales_as_at_22_Dec_2016.csv")
e
import matplotlib.pyplot as plt

# Drop rows where 'Global_Sales' is NaN
e = e.dropna(subset=['Global_Sales'])

# Calculate mean, sum, and range
mean_sales = e['Global_Sales'].mean()
sum_sales = e['Global_Sales'].sum()
sales_range = e['Global_Sales'].max() - e['Global_Sales'].min()

# Plotting
fig, ax = plt.subplots(figsize=(6, 6))

# Bar plot for mean
ax.bar('Mean', mean_sales, color='blue', label='Mean')

# Bar plot for sum
ax.bar('Sum', sum_sales, color='green', label='Sum')

# Bar plot for range
ax.bar('Range', sales_range, color='red', label='Range')

ax.set_ylabel('Global Sales (in millions)')
ax.set_title('Statistical Information about Global Sales')
ax.legend()

# Show the plot
plt.show()

```
